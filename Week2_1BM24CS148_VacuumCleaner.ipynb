```python
def reflex_agent(location, status):
    if status == 'dirty':
        return 'Suck'
    elif location == 'A':
        return 'Right'
    elif location == 'B':
        return 'Left'

environment = {
    "A": "dirty",
    "B": "dirty"
}


location = "A"


for step in range(4):
    status = environment[location]
    action = reflex_agent(location, status)

    print(f"Step {step + 1}: Location = {location}, Status = {status} -> Action = {action}")
    if action == 'Suck':
        environment[location] = 'clean'
    elif action == 'Right':
        location = 'B'
    elif action == 'Left':
        location = 'A'
```

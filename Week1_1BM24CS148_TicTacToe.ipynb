```python
def print_board(board):
    for row in board:
        print(" | ".join(row))
        print("---------")

def check_win(board, player):
    for row in board:
        if all([s == player for s in row]):
            return True
    for col in range(3):
        if all([board[row][col] == player for row in range(3)]):
            return True
    if all([board[i][i] == player for i in range(3)]) or \
       all([board[i][2 - i] == player for i in range(3)]):
        return True
    return False

def check_draw(board):
    for row in board:
        for spot in row:
            if spot == ' ':
                return False
    return True

def tic_tac_toe():
    board = [[' ' for _ in range(3)] for _ in range(3)]

    print("Welcome to Tic-Tac-Toe!")

    player1_name = input("Player 1, enter your name: ")
    player2_name = input("Player 2, enter your name: ")

    player1_choice_symbol = ''
    while player1_choice_symbol not in ['X', 'O']:
        player1_choice_symbol = input(f"{player1_name}, do you want to be X or O? ").upper()
        if player1_choice_symbol not in ['X', 'O']:
            print("Invalid choice. Please enter 'X' or 'O'.")

    if player1_choice_symbol == 'X':
        player_x_name = player1_name
        player_o_name = player2_name
    else:
        player_o_name = player1_name
        player_x_name = player2_name


    current_player_symbol = player1_choice_symbol

    game_over = False

    while not game_over:
        print_board(board)

        current_player_display_name = player_x_name if current_player_symbol == 'X' else player_o_name
        print(f"{current_player_display_name}'s turn (playing as {current_player_symbol}).")

        try:
            row = int(input("Enter row (1, 2, or 3): ")) - 1
            col = int(input("Enter column (1, 2, or 3): ")) - 1
        except ValueError:
            print("Invalid input. Please enter a number.")
            continue

        if not (0 <= row <= 2 and 0 <= col <= 2):
            print("Invalid row or column. Please enter numbers between 1 and 3.")
            continue

        if board[row][col] == ' ':
            board[row][col] = current_player_symbol

            if check_win(board, current_player_symbol):
                print_board(board)
                print(f"{current_player_display_name} wins!")
                game_over = True
            elif check_draw(board):
                print_board(board)
                print("It's a draw!")
                game_over = True
            else:

                current_player_symbol = 'O' if current_player_symbol == 'X' else 'X'
        else:
            print("That spot is already taken. Try again.")

    print("Game over.")

tic_tac_toe()
```

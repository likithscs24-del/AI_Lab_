```python
def dfs(state, goal, depth, limit, visited):
  if state == goal:
    print("\nFinal Solution Found!")
    for row in state:
      print(row)
    print(f"\nTotal number of moves: {depth}")
    return True

  if depth >= limit:
    return False

  if state in visited:
    return False
  visited.append(state)

  r, c = -1, -1
  for i in range(len(state)):
    for j in range(len(state[i])):
      if state[i][j] == 0:
        r, c = i, j
        break
    if r != -1:
      break

  moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
  for dr, dc in moves:
    new_r, new_c = r + dr, c + dc
    if 0 <= new_r < 3 and 0 <= new_c < 3:
      new_state = [list(row) for row in state]
      new_state[r][c], new_state[new_r][new_c] = new_state[new_r][new_c], new_state[r][c]

      if new_state not in visited:
        if dfs(new_state, goal, depth + 1, limit, visited):
          return True

  visited.remove(state)
  return False


def dls(intial, goal):
  print("Initial State:")
  for row in intial:
    print(row)

  limit = 0

  while True:

    visited = []

    result = dfs(intial, goal, 0, limit, visited)


    if result == True:
      print(f"\nGoal found within depth limit of {limit}.")
      break

    limit += 1



intial = [[2,8,3],
          [1,6,4],
          [7,0,5]]
goal = [[1,2,3],
        [8,0,4],
        [7,6,5]]
dls(intial,goal)
```

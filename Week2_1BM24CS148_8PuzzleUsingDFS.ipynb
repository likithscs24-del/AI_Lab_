```python
def dfs(intial,goal):
  print("Initial State:")
  for row in intial:
    print(row)
  stack = []
  visited = []
  stack.append([intial])

  while stack:
    path = stack.pop()
    state = path[-1]

    if(state == goal):
      print("Final Solution:")
      for row in state:
        print(row)
      print(f"Total number of moves: {len(path) - 1}")
      break

    if state in visited:
      continue
    visited.append(state)

    r,c = -1,-1
    for i in range(len(state)):
      for j in range(len(state[i])):
        if state[i][j] == 0:
          r,c = i,j
          break
      if r != -1:
        break

    moves = [(-1,0),(1,0),(0,-1),(0,1)]
    for dr,dc in moves:
      new_r,new_c = r + dr,c + dc
      if 0<= new_r < 3 and 0<= new_c < 3:
        new_state = [list(row) for row in state]
        new_state[r][c],new_state[new_r][new_c] = new_state[new_r][new_c],new_state[r][c]
        if new_state not in visited:
          stack.append(path + [new_state])


intial = [[2,8,3],
          [1,6,4],
          [7,0,5]]
goal = [[1,2,3],
        [8,0,4],
        [7,6,5]]
dfs(intial,goal)
```

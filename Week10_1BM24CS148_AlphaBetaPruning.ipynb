```python

def alpha_beta(values, depth):
    visited = []
    pruned = []

    def minimax(level, start, is_max, alpha, beta):
        # Terminal node
        if level == depth:
            visited.append(start + 1)
            return values[start], [values[start]], ""

        nodes_per_subtree = 2 ** (depth - level - 1)

        # Evaluate left child
        left_value, left_path, left_moves = minimax(
            level + 1, start, not is_max, alpha, beta
        )

        if is_max:
            best_value = left_value
            best_path = left_path
            best_moves = "L" + left_moves
            alpha = max(alpha, best_value)
        else:
            best_value = left_value
            best_path = left_path
            best_moves = "L" + left_moves
            beta = min(beta, best_value)

        # Alpha-Beta pruning condition
        if beta <= alpha:
            right_start = start + nodes_per_subtree
            pruned.extend(
                range(right_start + 1,
                      right_start + nodes_per_subtree + 1)
            )
            return best_value, [best_value] + best_path, best_moves

        # Evaluate right child
        right_value, right_path, right_moves = minimax(
            level + 1,
            start + nodes_per_subtree,
            not is_max,
            alpha,
            beta
        )

        # Select maximum or minimum
        if is_max:
            if right_value > best_value:
                best_value = right_value
                best_path = right_path
                best_moves = "R" + right_moves
        else:
            if right_value < best_value:
                best_value = right_value
                best_path = right_path
                best_moves = "R" + right_moves

        return best_value, [best_value] + best_path, best_moves

    optimal_value, path, moves = minimax(
        0, 0, True, float('-inf'), float('inf')
    )

    return optimal_value, path, moves, visited, pruned


# Main program
print("MINMAX WITH ALPHA-BETA PRUNING")
print("--------------------------------")

depth = int(input("Enter the depth of the tree: "))

if depth < 1:
    print("Depth must be at least 1.")
else:
    total_nodes = 2 ** depth

    print("\nNumber of terminal nodes:", total_nodes)

    print("Enter", total_nodes,
          "terminal node values separated by spaces:")

    values = list(map(int, input().split()))

    if len(values) != total_nodes:
        print("Error! Please enter exactly",
              total_nodes, "values.")
    else:
        result, path, moves, visited, pruned = alpha_beta(
            values, depth
        )

        print("\n------ RESULT ------")
        print("Optimal Value:", result)
        print("Optimal Path:", " -> ".join(map(str, path)))

        print("Path Direction:", " -> ".join(moves)
              if moves else "Root")

        print("Visited Terminal Nodes:", visited)
        print("Pruned Terminal Nodes:", pruned)

        print("Total Terminal Nodes:", total_nodes)
        print("Visited Nodes:", len(visited))
        print("Pruned Nodes:", len(pruned))
```

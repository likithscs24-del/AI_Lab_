```python
import math
import random


class SimulatedAnnealing8Queens:

    def __init__(self, initial_state):
        # 1. State Representation: 1D array of length 8 where board[col] = row
        self.initial_state = list(initial_state)

    def calculate_cost(self, state):
        """2. Cost Function (Objective): Count all unique pair conflicts."""
        conflicts = 0
        n = len(state)
        for c1 in range(n):
            for c2 in range(c1 + 1, n):
                r1 = state[c1]
                r2 = state[c2]

                # Check for Row Collision or Diagonal Collision
                if r1 == r2 or abs(r1 - r2) == abs(c1 - c2):
                    conflicts += 1
        return conflicts

    def get_random_neighbor(self, state):
        """3. Neighbor Generation: Pick a random column and change its row value."""
        neighbor = list(state)
        col = random.randint(0, 7)

        # Pick a different random row from 0 to 7
        current_row = state[col]
        possible_rows = [r for r in range(8) if r != current_row]
        new_row = random.choice(possible_rows)

        neighbor[col] = new_row
        return col, new_row, neighbor

    def run_simulation(
        self, initial_temp=10.0, alpha=0.995, max_iterations=10000
    ):
        current_board = list(self.initial_state)
        current_cost = self.calculate_cost(current_board)
        T = initial_temp
        iteration = 0

        print("\n=== STARTING SIMULATED ANNEALING ===")
        print(
            f"Initial Configuration: {current_board} | Cost: {current_cost} conflicts\n"
        )

        while T > 0.0001 and current_cost > 0 and iteration < max_iterations:
            iteration += 1
            col, new_row, proposed_board = self.get_random_neighbor(
                current_board
            )
            proposed_cost = self.calculate_cost(proposed_board)

            # 4. Acceptance Criterion: delta_E = Cost_new - Cost_current
            delta_E = proposed_cost - current_cost

            # Case 1: delta_E < 0 (Better Move) -> Always Accept
            if delta_E < 0:
                current_board = proposed_board
                current_cost = proposed_cost

            # Case 2: delta_E >= 0 (Worse Move) -> Accept Probabilistically
            else:
                p = math.exp(-delta_E / T)
                if random.random() < p:
                    current_board = proposed_board
                    current_cost = proposed_cost

            # 5. Cooling Schedule: T_(k+1) = T_k * alpha
            T *= alpha

        print("=== CONVERGENCE SUMMARY ===")
        print(f"Total Iterations Required: {iteration}")
        if current_cost == 0:
            print(f"SUCCESS: Found goal state: {current_board}")
        else:
            print(
                f"STOPPED: Local minimum reached: {current_board} | Remaining Cost: {current_cost}"
            )

        self.print_board(current_board)

    def print_board(self, state):
        """Draws a visual text-based 8x8 chessboard matrix."""
        print("\nVisual Board Layout:")
        for row in range(8):
            line = ""
            for col in range(8):
                line += " Q " if state[col] == row else " . "
            print(line)


def get_user_board():
    """Gets and validates custom initial state input from the user."""
    print("=== 8-QUEENS SIMULATED ANNEALING SETUP ===")
    print("Choose an option for the starting board state:")
    print("1. Enter 8 space-separated integers (0-7 for row positions)")
    print("2. Generate a random starting board")
    print("3. Exit")

    choice = input("\nEnter choice (1, 2, or 3): ").strip()

    if choice == "1":
        while True:
            try:
                user_input = input(
                    "Enter 8 row numbers between 0 and 7 (e.g., 0 4 7 5 2 6 1 3): "
                )
                board = [int(x) for x in user_input.split()]

                if len(board) != 8:
                    print("Error: You must enter exactly 8 numbers.")
                    continue

                if any(r < 0 or r > 7 for r in board):
                    print("Error: All row values must be between 0 and 7.")
                    continue

                return board
            except ValueError:
                print("Error: Invalid input. Please enter integers only.")

    elif choice == "2":
        random_board = [random.randint(0, 7) for _ in range(8)]
        print(f"Generated Random Board: {random_board}")
        return random_board

    else:
        print("Enter Wrong Option Exit")


if __name__ == "__main__":
    initial_board = get_user_board()
    sa = SimulatedAnnealing8Queens(initial_board)
    sa.run_simulation()
```

```python
import random

def calculate_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor_permutation(state, size=4):
    current_conflicts = calculate_conflicts(state)
    best_neighbors = []
    best_conflicts = current_conflicts

    for i in range(size):
        for j in range(i + 1, size):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbor_conflicts = calculate_conflicts(neighbor)

            if neighbor_conflicts < best_conflicts:
                best_conflicts = neighbor_conflicts
                best_neighbors = [neighbor]
            elif neighbor_conflicts == best_conflicts:
                best_neighbors.append(neighbor)

    if best_neighbors:
        best_neighbors.sort()
        chosen = best_neighbors[-1]
        print(f"Available best neighbors {best_neighbors} with conflicts ({best_conflicts})")
        return chosen, best_conflicts
    return state, current_conflicts

def display_board(state, size=4):
    for row in range(size):
        line = ""
        for col in range(size):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
    print("-" * 15)

def solve_hill_climbing_permutation():
    size = 4
    state = [3, 1, 2, 0]

    print(f"Initial State: {state}")
    print("Initial Board Layout:")
    display_board(state, size)

    steps = 0
    max_steps = 100

    while steps < max_steps:
        current_conflicts = calculate_conflicts(state)

        if current_conflicts == 0:
            print(f"Solution found! Total Steps taken: {steps}\n")
            print(f"Final State: {state}")
            display_board(state, size)
            return state

        next_state, next_conflicts = get_best_neighbor_permutation(state, size)
        steps += 1

        if next_state == state:
            state = [0, 1, 2, 3]
            random.shuffle(state)
            continue

        state = next_state
        print(f"Step {steps}: Selected and moved to state {state} (Conflicts: {next_conflicts})\n")

if __name__ == "__main__":
    solve_hill_climbing_permutation()
```
